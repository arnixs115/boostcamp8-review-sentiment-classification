```python
import pandas as pd

# 1. 테스트 데이터 로드
test_df = pd.read_csv("여러분의 test.csv 경로")

# 2. 텍스트 컬럼만 추출 (예: 'review' 컬럼)
text_data = test_df['review']
text_data = text_data.fillna('').astype('str')
```

```python
# 3. .txt 파일로 저장 (한 줄에 한 문장)
with open("tapt_corpus.txt", "w", encoding="utf-8") as f:
    for line in text_data:
        f.write(line + "\n")
        
print("TAPT용 텍스트 파일 'tapt_corpus.txt' 생성 완료.")
```

```python
!python run_mlm.py \
    --model_name_or_path "klue/roberta-base" \
    --train_file 'data/tapt_corpus.txt' \
    --do_train \
    --line_by_line True \
    --num_train_epochs 2 \
    --per_device_train_batch_size 16 \
    --output_dir /data/ephemeral/home/tapted-bert-model2 \
    --fp16 True \
    --save_strategy "steps" \
    --save_steps 500 \
    --save_total_limit 3
```
